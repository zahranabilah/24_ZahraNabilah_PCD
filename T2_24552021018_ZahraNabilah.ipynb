### **Tugas 2 — Mengolah String**

Buat program untuk mengolah data nama dan kalimat dengan ketentuan:

- Simpan nama depan dan nama belakang dalam variabel.
- Gabungkan keduanya menjadi nama lengkap.
- Tampilkan sebuah kalimat menggunakan `upper()`, `lower()`, `capitalize()`, dan `title()`.
- Hitung jumlah kemunculan huruf `a` menggunakan `count()`.
- Tampilkan hasil setiap operasi.

Output yang diharapkan: nama lengkap, perubahan format kalimat, dan jumlah karakter.



````python
# Tugas 2 — Mengolah String

# 1. Menyimpan nama depan dan nama belakang dalam variabel
nama_depan = "Zahra"
nama_belakang = "Nabilah"

# 2. Menggabungkan keduanya menjadi nama lengkap
nama_lengkap = nama_depan + " " + nama_belakang
print("Nama Lengkap:", nama_lengkap)

# 3. Membuat sebuah kalimat untuk diolah
kalimat_asli = "belajar pemrograman python mulai dari nol!"
print("Kalimat Asli:", kalimat_asli)
print("-----------------------------------------------------------------")

# 4. Mengubah format kalimat menggunakan berbagai metode string
kalimat_upper = kalimat_asli.upper()           # Mengubah menjadi huruf kapital semua
kalimat_lower = kalimat_asli.lower()           # Mengubah menjadi huruf kecil semua
kalimat_capitalize = kalimat_asli.capitalize() # Mengubah huruf pertama kalimat menjadi kapital
kalimat_title = kalimat_asli.title()           # Mengubah huruf pertama setiap kata menjadi kapital

# Menampilkan hasil operasi format string
print("Hasil upper()      :", kalimat_upper)
print("Hasil lower()      :", kalimat_lower)
print("Hasil capitalize() :", kalimat_capitalize)
print("Hasil title()      :", kalimat_title)
print("-----------------------------------------------------------------")

# 5. Menghitung jumlah kemunculan huruf 'a' menggunakan count()
jumlah_huruf_a = kalimat_asli.count('a')
print(f"Jumlah kemunculan huruf 'a' pada kalimat asli: {jumlah_huruf_a} kali")
````

### **Tugas 3 — Operasi Matematika**

Buat program kalkulator sederhana dengan ketentuan:

- Minta pengguna memasukkan dua angka.
- Konversikan input menjadi `int` atau `float`.
- Hitung penjumlahan, pengurangan, perkalian, pembagian, dan pangkat.
- Tampilkan hasil setiap operasi.
- Tambahkan perhitungan luas persegi panjang berdasarkan input panjang dan lebar.

Output yang diharapkan: hasil kalkulasi berdasarkan angka yang dimasukkan pengguna.



````python
# Tugas 3 — Operasi Matematika

# BAGIAN 1: Kalkulator Sederhana
print("=====================================")
print("        KALKULATOR SEDERHANA         ")
print("=====================================")

# Menerima input dua angka dari pengguna dan mengonversinya menjadi float
angka1 = float(input("Masukkan angka pertama: "))
angka2 = float(input("Masukkan angka kedua: "))
print("-------------------------------------")

# Melakukan perhitungan matematika dasar
penjumlahan = angka1 + angka2
pengurangan = angka1 - angka2
perkalian = angka1 * angka2

# Memastikan tidak terjadi pembagian dengan nol
if angka2 != 0:
    pembagian = angka1 / angka2
else:
    pembagian = "Tidak terdefinisi (pembagian dengan nol)"

pangkat = angka1 ** angka2

# Menampilkan hasil kalkulasi
print(f"Hasil Penjumlahan ({angka1} + {angka2}) : {penjumlahan}")
print(f"Hasil Pengurangan ({angka1} - {angka2}) : {pengurangan}")
print(f"Hasil Perkalian   ({angka1} * {angka2}) : {perkalian}")
print(f"Hasil Pembagian   ({angka1} / {angka2}) : {pembagian}")
print(f"Hasil Pangkat     ({angka1} ^ {angka2}) : {pangkat}")
print("=====================================\n")


# --- BAGIAN 2: Menghitung Luas Persegi Panjang ---
print("=====================================")
print("    HITUNG LUAS PERSEGI PANJANG      ")
print("=====================================")

# Menerima input panjang dan lebar kemudian dikonversi ke float
panjang = float(input("Masukkan panjang persegi panjang: "))
lebar = float(input("Masukkan lebar persegi panjang  : "))

# Rumus luas persegi panjang
luas = panjang * lebar

# Menampilkan hasil perhitungan luas
print("-------------------------------------")
print(f"Luas Persegi Panjang adalah: {luas}")
print("=====================================")
````

### **Tugas 4 — Perulangan `for` dan `while`**

Buat program untuk menampilkan data mahasiswa dengan ketentuan:

- Buat daftar minimal lima nama mahasiswa.
- Tampilkan semua nama menggunakan `for`.
- Tampilkan angka 1 sampai 10 menggunakan `range()`.
- Tampilkan kembali daftar nama menggunakan `while` dan `len()`.
- Buat perulangan `while` untuk menampilkan bilangan genap dari 2 sampai 20.

Output yang diharapkan: daftar nama dan deret angka yang tampil sesuai ketentuan.


````python
# Tugas 4 — Perulangan for dan while

# 1. Membuat daftar (list) dengan 5 nama mahasiswa
mahasiswa = ["Zahra", "Budi", "Andi", "Salma", "Rian"]

# 2. Menampilkan semua nama menggunakan perulangan 'for'
print("=== DAFTAR MAHASISWA (Menggunakan Perulangan for) ===")
for nama in mahasiswa:
    print(f"Nama: {nama}")
print("---------------------------------------------------\n")

# 3. Menampilkan angka 1 sampai 10 menggunakan range() dan perulangan for
print("=== DERET ANGKA 1 SAMPAI 10 (Menggunakan range) ===")
for i in range(1, 11):
    print(i, end=" ")
print("\n---------------------------------------------------\n")

# 4. Menampilkan kembali daftar nama menggunakan perulangan 'while' dan fungsi len()
print("=== DAFTAR MAHASISWA (Menggunakan Perulangan while & len) ===")
i = 0
while i < len(mahasiswa):
    print(f"Mahasiswa ke-{i + 1}: {mahasiswa[i]}")
    i += 1
print("---------------------------------------------------\n")

# 5. Membuat perulangan 'while' untuk menampilkan bilangan genap dari 2 sampai 20
print("=== BILANGAN GENAP 2 SAMPAI 20 (Menggunakan while) ===")
angka = 2
while angka <= 20:
    print(angka, end=" ")
    angka += 2
print("\n---------------------------------------------------")
````

**Tugas 5 — Fungsi dan Waktu**

Buat program pencatatan aktivitas belajar dengan ketentuan:

- Impor modul `datetime`.
- Buat fungsi bernama `catat_waktu()`.
- Tampilkan pesan `Aktivitas selesai`.
- Tampilkan waktu saat fungsi dijalankan menggunakan `datetime.datetime.now()`.
- Panggil fungsi sebelum dan sesudah perulangan.
- Gunakan perulangan `for` untuk menampilkan proses pengerjaan lima soal.

Output yang diharapkan: pesan aktivitas, waktu pelaksanaan, dan proses pengerjaan soal.




````python
# Tugas 5 — Fungsi dan Waktu

# 1. Mengimpor modul datetime untuk mengolah waktu
import datetime

# 2. Membuat fungsi catat_waktu()
def catat_waktu():
    # Mendapatkan waktu saat ini
    waktu_sekarang = datetime.datetime.now()
    # Menampilkan pesan dan waktu pelaksanaan
    print("=========================================")
    print("Aktivitas Selesai!")
    print(f"Waktu Pelaksanaan: {waktu_sekarang}")
    print("=========================================\n")

# 3. Memanggil fungsi sebelum perulangan dimulai
print("--- Mencatat Waktu Mulai Aktivitas ---")
catat_waktu()

# 4. Menggunakan perulangan for untuk menampilkan proses pengerjaan lima soal
print("--- PROSES PENGERJAAN SOAL ---")
for i in range(1, 6):
    print(f"Sedang mengerjakan Soal nomor {i}...")
print("-----------------------------------------\n")

# 5. Memanggil fungsi sesudah perulangan selesai
print("--- Mencatat Waktu Selesai Aktivitas ---")
catat_waktu()
````
