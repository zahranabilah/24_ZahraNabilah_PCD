**Soal 2 (NumPy - Cropping):**
*   Lakukan pemotongan (cropping) gambar pada koordinat baris/y $200$ hingga $600$ dan kolom/x $400$ hingga $800$ menggunakan slicing array NumPy, lalu tampilkan hasilnya.



```python
## Jawaban Soal No 2
import cv2
import matplotlib.pyplot as plt

# Membaca gambar menggunakan OpenCV
img_bgr = cv2.imread('orange.jpg')

# Mengubah ruang warna BGR ke RGB agar sesuai dengan tampilan asli di Matplotlib
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Melakukan pemotongan (cropping) menggunakan slicing array NumPy
# Koordinat baris (y): 200 hingga 600, Kolom (x): 400 hingga 800
cropped_img = img_rgb[200:600, 400:800]

# Menampilkan hasil pemotongan
plt.figure(figsize=(6, 6))
plt.imshow(cropped_img)
plt.title("Hasil Cropping (y: 200-600, x: 400-800)")
plt.axis('off')
plt.show()
```

**Soal 3 (Scikit-Image - Filtering):**
*   Konversi hasil potongan (crop) dari Soal 2 menjadi grayscale, lalu terapkan filter deteksi tepi Sobel (`filters.sobel`) dari Scikit-Image dan tampilkan hasilnya dengan peta warna (cmap) abu-abu (`gray`).

```python
## Jawaban Soal No 3
import cv2
from skimage import filters
import matplotlib.pyplot as plt

# 1. Membaca gambar dan melakukan pemotongan seperti pada Soal 2
img_bgr = cv2.imread('orange.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
cropped_img = img_rgb[200:600, 400:800]

# 2. Mengonversi hasil pemotongan (crop) ke grayscale
cropped_gray = cv2.cvtColor(cropped_img, cv2.COLOR_RGB2GRAY)

# 3. Menerapkan filter deteksi tepi Sobel dari Scikit-Image
edges = filters.sobel(cropped_gray)

# 4. Menampilkan hasil dengan peta warna (cmap) abu-abu (gray)
plt.figure(figsize=(6, 6))
plt.imshow(edges, cmap='gray')
plt.title("Hasil Deteksi Tepi Sobel (Cropped Image)")
plt.axis('off')
plt.show()
```
